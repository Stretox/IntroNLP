## Understanding TF-IDF

Counting word occurrences is useful, but it has a limitation: very frequent words (like "the", "is", "a") dominate the representation even though they may not be very informative.

**TF-IDF (Term Frequency-Inverse Document Frequency)** addresses this by:
- Weighing words by how important they are to a specific document
- Penalizing words that appear in many documents

### The Math Behind TF-IDF

**TF-IDF** = TF × IDF

- **TF (Term Frequency)**: \\(\\log(count(t,d) + 1)\\)
  - How often a term \\(t\\) appears in document \\(d\\)
  
- **IDF (Inverse Document Frequency)**: \\(\\log\\frac{|D|}{\\# d : term \\in doc}\\)
  - \\(|D|\\) = total number of documents
  - Denominator = number of documents containing the term
  - Rare words get higher IDF scores

**Example:**
- If "Trump" appears in only 10 out of 1000 documents, it gets a high IDF
- If "the" appears in 999 out of 1000 documents, it gets a very low IDF

We calculate TF-IDF for each term in each document.

## Introduction to Word Embeddings

So far we've represented words as sparse vectors (bag-of-words, TF-IDF). These representations have limitations:
- **High dimensionality**: Vocabulary size can be tens of thousands of features
- **Sparsity**: Most values are zero
- **No semantic meaning**: "cat" and "dog" are just as different as "cat" and "car"

**Word embeddings** offer a different approach:

- **Dense vectors**: Each word is represented by a vector of typically 50-300 dimensions (not thousands!)
- **Semantic meaning**: Words with similar meanings have similar vectors
- **Learned from data**: Trained on large corpora to capture word relationships

### Famous Word Embedding Models:

- **Word2Vec** (Google, 2013): Predicts context from words or words from context
- **GloVe** (Stanford): Global vectors for word representation
- **FastText** (Facebook): Includes subword information

### Why are embeddings powerful?

They capture semantic relationships:
- **Similarity**: vector("king") ≈ vector("queen"), both are royalty
- **Analogies**: vector("king") - vector("man") + vector("woman") ≈ vector("queen")
- **Synonyms**: "car" and "automobile" have similar vectors

- Easy access to pre-trained Word2Vec, GloVe, and FastText models
